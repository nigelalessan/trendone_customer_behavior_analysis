## 4. Data Cleaning

### Standardize columns name

```python
# Standardize columns name

df_prep.columns = df_prep.columns.str.lower()
df_prep.columns = df_prep.columns.str.replace(' ', '_')

df_prep = df_prep.rename(columns = {'purchase_amount_(usd)' : 'purchase_amount_usd'})
df_prep.head()
```

The column names shown above have been updated to follow standard naming conventions.

### Missing values data imputation

```python
# Missing values data imputation
df_prep.info()
```

```python
df_prep['review_rating'] = df_prep['review_rating'].fillna(df_prep['review_rating'].mean().round(1))
df_prep.info()
```

Missing values in the review_rating column have been imputed using the mean, leaving no null values remaining in the dataset.

## 5. Feature Engineering

### Age group
`Age Group` is created to support customer-level comparisons across age ranges.

The feature is derived from `Age` and will be used as a grouping variable in subsequent customer analysis. The original `Age` variable is retained to preserve the underlying numerical information.

```python
bins = [17, 24, 34, 44, 54, 64, 70]
labels = ["18-24", "25-34", "35-44", "45-54", "55-64", "65-70"]

df_prep['age_group'] = pd.cut(
    df_prep["age"],
    bins=bins,
    labels=labels
)
```

I created these bins by taking into account the minimum and maximum age values, ensuring they focus strictly on the observed age range.

### Previous Purchases Group

```python
bins=[-1, 5, 10, 20, float("inf")]
labels=["0-5", "6-10", "11-20", "21+"]

df_prep['previous_purchases_group'] = pd.cut(
    df_prep["previous_purchases"],
    bins=bins,
    labels=labels
)
```

I grouped the previous_purchases data to identify frequent buyers, which can serve as a valuable analytical insight.
