### 5.4 Potential Outliers

Potential outliers are assessed to identify observations that are unusually distant from the rest of the distribution.

An outlier is not automatically treated as an error. Any potential outlier identified at this stage is flagged for further investigation
rather than removed or modified.

```python
def iqr_outlier_summary(df, columns):
    results = []

    for col in columns:
        q1 = df[col].quantile(0.25)
        q3 = df[col].quantile(0.75)
        iqr = q3 - q1

        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr

        outliers = ((df[col] < lower) | (df[col] > upper)).sum()

        results.append({
            "Column": col,
            "Q1": q1,
            "Q3": q3,
            "IQR": iqr,
            "Lower Bound": lower,
            "Upper Bound": upper,
            "Potential Outliers": outliers
        })

    return pd.DataFrame(results)

iqr_outlier_summary(df, numerical_cols)
```

As outlined in the rough overview in section 5.3, it has been verified that the dataset contains no extreme outliers across 
its numerical columns.
